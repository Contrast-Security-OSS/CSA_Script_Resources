# Full Application Assessment Report Notebook

Notebook wrapper for `Reports/full-app-report/generate_app_report.py`.

Outputs are written under `notebooks/full-app-report/Output/`.

## Flow
1. Set env vars manually (Cell 3) or load root `.env` (Cell 4).
2. Configure runtime paths (Cell 5).
3. Set selection parameters (Cell 6).
4. Run the script (Cell 7).

```python
import os

os.environ["TEAMSERVER_URL"] = "https://your_saas_instance.contrastsecurity.com/"
os.environ["ORG_UUID"] = ""
os.environ["CONTRAST_AUTH"] = ""
os.environ["CONTRAST_API_KEY"] = ""
```

```python
from dotenv import load_dotenv
load_dotenv(dotenv_path="../../.env", override=True)
print("Loaded .env from ../../.env")
```

```python
from pathlib import Path
from datetime import datetime

def find_repo_root(start):
    current = Path(start).resolve()
    for p in [current, *current.parents]:
        if (p / "notebooks").exists() and (p / "README.md").exists():
            return p
    return current

repo_root = find_repo_root(Path.cwd())
script_path = repo_root / "Reports" / "full-app-report" / "generate_app_report.py"
output_dir = repo_root / "notebooks" / "full-app-report" / "Output"
date_str = datetime.now().strftime("%Y-%m-%d")
print("Runtime configuration loaded.")
print(f"Script: {script_path}")
print(f"Output directory: {output_dir}")
```

```python
# Set one selector: app_name or tag
app_name = ""
tag = ""
customer_name = "Customer"
days = 365
skip_staleness = False
skip_servers = False

if not app_name and not tag:
    raise ValueError("Set app_name or tag before running the next cell.")

label = app_name if app_name else tag
safe_label = label.replace(" ", "_").replace("/", "_")
output_path = output_dir / f"{customer_name.replace(' ', '_')}_{safe_label}_{date_str}.md"
print(f"Output file: {output_path}")
```

```python
import subprocess
import sys

cmd = [
    sys.executable, str(script_path),
    "--output", str(output_path),
    "--customer-name", customer_name,
    "--days", str(days)
]
if app_name:
    cmd += ["--app-name", app_name]
if tag:
    cmd += ["--tag", tag]
if skip_staleness:
    cmd += ["--skip-staleness"]
if skip_servers:
    cmd += ["--skip-servers"]

result = subprocess.run(cmd, cwd=repo_root, capture_output=True, text=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(f"Script failed with exit code {result.returncode}")
print(f"Done: {output_path}")
```
